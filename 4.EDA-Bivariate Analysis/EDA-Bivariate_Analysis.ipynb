### 3.2.2 Multicollinearity - VIF

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
def calc_vif(X):
    vif=pd.DataFrame()
    vif["variables"]=X.columns
    vif["VIF"]=[variance_inflation_factor(X.values,i) for i in range(X.shape[1])]
    return(vif)
```

```python
calc_vif(dataset[["length_url","length_hostname","nb_dots","nb_hyphens","nb_at"]])
```

### 3.2.3 Hypothesis Testing

#### 3.2.3.1 T-Test

```python
from scipy.stats import ttest_ind
phishing=dataset[dataset["status"]==1]["length_url"]
legitimate=dataset[dataset["status"]==0]["length_url"]
ttest_ind(phishing,legitimate)
```
